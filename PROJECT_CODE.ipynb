```python
# Libraries
import dash
from dash import html, dcc, callback, Input, Output
import pandas as pd
import plotly.express as px
```

```python
# load all the dataset for each Province and Canada
data = {
    "Canada": "Canada.csv",
    "Alberta": "Alberta.csv",
    "British Columbia": "British_Columbia.csv",
    "Manitoba": "Manitoba.csv",
    "New Brunswick": "New_brunswick.csv",
    "Newfoundland": "Newfoundland_labrador.csv",
    "North West Territories": "North_West_Territories.csv",
    "Nova Scotia": "Nova_scotia.csv",
    "Nunavut": "Nunavut.csv",
    "Ontario": "Ontario.csv",
    "Prince Edward Island": "Prince_Edward_Island.csv",
    "Quebec": "Quebec.csv",
    "Saskatchewan": "Saskatchewan.csv",
    "Yukon": "Yukon.csv"
}

# Clean the data for readability
def clean_data(fp, province):

    # read the data 
    df = pd.read_csv(fp, encoding="utf-8")
    
    # rename columns
    df.columns = ["Occupation", "Total", "Men", "Women"]
    
    # loop through each column and clean it (remove commas, extract numbers, fill in empty cells (i.e. NA), and convert to integers
    for col in ["Total", "Men", "Women"]:
        df[col] = df[col].astype(str).str.replace(",", "").str.extract(r"(\d+)").fillna(0).astype(int)
        
    # add a Province column to the new data
    df["Province"] = province
    
    return df

# load and merge each dataframe together into a single one 
df_list = [clean_data(path, province) for province, path in data.items()]
master_df = pd.concat(df_list)

# extracts the NOC level (1-9)
master_df['NOC'] = master_df['Occupation'].str.extract(r'^(\d)')

# Initialize the Dash app
app = dash.Dash(__name__)

# define the layout of dashboard
app.layout = html.Div([

    # Title of Dashboard
    html.H1("CP321 W25 - Final Project", style={"textAlign": "center"}),

    # Line for organization
    html.Hr(style={"margin": "40px 0"}),

# --------------------------------------------------------------------------------------------
# Visualization - QUESTION 1 
# --------------------------------------------------------------------------------------------
     html.Div([

        # header for q1
        html.H2("Q1 - Distribution of Essential Services in Canada", style={"textAlign": "center"}),
    
        html.Div([
            
            html.Div([
                # label for menu
                html.Label("Select a Province/Territory:", style={'fontWeight': 'bold', 'fontSize': '18px', 'marginBottom': '10px'}),
                
                # widget that provide a list of the region (Canada, Provinces, Territories), start value is Canada
                dcc.RadioItems(
                    id='Province/Territory_ES', 
                    options=[{'label': k, 'value': k} for k in data.keys()], 
                    value='Canada', 
                    labelStyle={'display': 'block', 'fontSize': '16px', 'margin': '4px 0'}),
            ], 
            # Style formatting, making the list and pie chart inline         
            style={'display': 'inline-block', 'width': '25%', 'verticalAlign': 'top', 'paddingRight': '20px', 'marginTop': '45px'}),
    
            html.Div([
                # displays the pie chart
                dcc.Graph(id='Essential_Service', style={'height': '500px'})
            ], 
            # Style formatting, making the list and pie chart inline         
            style={'display': 'inline-block', 'width': '70%', 'verticalAlign': 'top'}
            )
        ], 
        style={'maxWidth': '1000px', 'margin': '0 auto'}
        )
    ]),

    # Line for organization
    html.Hr(style={"margin": "40px 0"}),

    
# --------------------------------------------------------------------------------------------
# Visualizaon - QUESTION 2
# -------------------------------------------------------------------------------------------- 
    html.Div([

        # header for q2
        html.H2("Q2 - Employment by Gender in Canada", style={"textAlign": "center"}),

        # label for dropdown menu
        html.Label("Select Occupation", style={'fontWeight': 'bold', 'fontSize': '18px'}),

        # widget that provides a dropdown menu for each occupation, start value is 0
        dcc.Dropdown(
            id='Occupation',
            options=[
                {'label': "0 Legislative and senior management occupations", 'value': '0'},
                {'label': "1 Business, finance and administration occupations", 'value': '1'},
                {'label': "2 Natural and applied sciences and related occupations", 'value': '2'},
                {'label': "3 Health occupations", 'value': '3'},
                {'label': "4 Occupations in education, law and social, community and government services", 'value': '4'},
                {'label': "5 Occupations in art, culture, recreation and sport", 'value': '5'},
                {'label': "6 Sales and service occupations", 'value': '6'},
                {'label': "7 Trades, transport and equipment operators and related occupations", 'value': '7'},
                {'label': "8 Natural resources, agriculture and related production occupations", 'value': '8'},
                {'label': "9 Occupations in manufacturing and utilities", 'value': '9'}
            ],
            value='0',
            placeholder="Select a Occupation"
        ),

        # displays graph for each Province and Territories
        dcc.Graph(id='Provinces/Territories_GE', style={'height': '500px'}),
        # displays graph for Canada, for better interpretation
        dcc.Graph(id='Canada_GE', style={'height': '500px'}),
    ]),

    # Line for organization
    html.Hr(style={"margin": "40px 0"}),
    

# --------------------------------------------------------------------------------------------
# Visualization - QUESTION 3
# -------------------------------------------------------------------------------------------- 
    html.Div([

        # header for q3
        html.H2("Q3 - Engineering Occupations in Canada:", style={"textAlign": "center"}),
        # label for the checklist
        html.Label("Select Engineering Occupaton:", style={'fontWeight': 'bold', 'fontSize': '18px'}),

        # widget that provides a checklist of the types of Engineering Occupation
        dcc.Checklist(
            id='Engineering_Occupation',
            options=[
                {'label': 'Mechanical Engineers', 'value': 'mechanical'},
                {'label': 'Computer Engineers', 'value': 'computer'},
                {'label': 'Electrical Engineers', 'value': 'electrical'}
            ],
            inline=True,
            labelStyle={'marginRight': '15px'}
        ),
        
        html.Div([
            html.Div([
                # displays bar graph of Provinces and Territories
                dcc.Graph(id='Province/Territory_ENG')
            ], 
            # Style formatting, making both graphs inline               
            style={'display': 'inline-block', 'width': '70%'}
            ),
            
            html.Div([
                # displays bar graph of Canada
                dcc.Graph(id='Canada_ENG')
            ],
            # Style formatting, making both graphs inline 
            style={'display': 'inline-block', 'width': '30%'}
            )
        ])
    ]),

    # Line for organization
    html.Hr(style={"margin": "40px 0"}),

    
# --------------------------------------------------------------------------------------------
# Visualization - QUESTION 4
# -------------------------------------------------------------------------------------------- 
    html.Div([

        # header for q4
        html.H2("Q4 - Top Occupations in Canada", style={"textAlign": "center"}),

        # label for the dropdown menu
        html.Label("Select Province/Territory:", style={'fontWeight': 'bold', 'fontSize': '18px'}),

        # widget that provides a dropdown menu of the region (Canada, Provinces, Territories), start value is Canada
        dcc.Dropdown(
            id='Province/Territory_Top',
            options=[{'label': k, 'value': k} for k in data.keys()],
            value='Canada',
            placeholder="Select a Province/Territory"
        ),

        # label for the slider 
        html.Label("Select Number of Top Occupations to Display:", style={'fontWeight': 'bold', 'fontSize': '18px'}),

        # widget that provides a slider for the Top occupation (NOC 0-9) for the selected region 
        dcc.Slider(
            id='Slider',
            min=1,
            max=9,
            step=1,
            value=1,
            marks={i: str(i) for i in range(1, 10)}
        ),
        # displays pie chart
        dcc.Graph(id='Top_Occupation', style={'height': '600px'})
    ])

])


# --------------------------------------------------------------------------------------------
# Callback and Graph - QUESTION 1
# --------------------------------------------------------------------------------------------
# callback for q1 graph
@app.callback(
    Output('Essential_Service', 'figure'),
    Input('Province/Territory_ES', 'value')
)

def q1(region):

    # filters row based on selected region
    df = master_df[master_df['Province'] == region]
    # defines the essential services and list to hold totals
    essential_services = ['nurse', 'police', 'firefighter']
    values = []

    # Loop through each occupation and calculate the total # of workers
    for keyword in essential_services:
        count = df[df['Occupation'].str.contains(keyword, case=False)]['Total'].sum()
        values.append(count)

    # creates pie chart 
    fig = px.pie(names=[s.title() for s in essential_services], values=values, title=f"Essential Services in {region}",)
    # format layout for organization
    fig.update_layout(title_x=0.5, title_font_size=22)
    
    return fig


# --------------------------------------------------------------------------------------------
# Callback and Graph - QUESTION 2
# --------------------------------------------------------------------------------------------
# callback for q2 graph
@app.callback(
    Output('Provinces/Territories_GE', 'figure'),
    Output('Canada_GE', 'figure'),
    Input('Occupation', 'value')
)

def q2(NOC):
    
    # filters row based on NOC 
    df = master_df[master_df['NOC'] == NOC]

    # separates the provinces and territories data from Canada data
    provinces_df = df[df['Province'] != 'Canada']
    canada_df = df[df['Province'] == 'Canada']

    # Sums number of men and women per Province/Territory, and formats data to long format
    prov_terr_q1 = provinces_df.groupby('Province')[['Men', 'Women']].sum().reset_index()
    prov_terr_melted = pd.melt(prov_terr_q1, id_vars='Province', value_vars=['Men', 'Women'], var_name='Gender', value_name='Count')

    # creates horizontal bar chart for Province/Territory
    fig1 = px.bar(prov_terr_melted, y='Province', x='Count', color='Gender', barmode='group', title='Provinces and Territories')
    # format layout for organization
    fig1.update_layout(title_x=0.5, title_font_size=22, height=700)

    # Sums number of men and women in Canada, and formats data to long format
    can_q1 = canada_df.groupby('Province')[['Men', 'Women']].sum().reset_index()
    can_melted = pd.melt(can_q1, id_vars='Province', value_vars=['Men', 'Women'], var_name='Gender', value_name='Count')

    # creates horizontal bar chart for Canada
    fig2 = px.bar(can_melted, y='Gender', x='Count', color='Gender', title='Canada')
    # format layout for organization
    fig2.update_layout(title_x=0.5, title_font_size=22, height=300)

    return fig1, fig2


# --------------------------------------------------------------------------------------------
# Callback and Graph - QUESTION 3
# --------------------------------------------------------------------------------------------
# callback for q3 graph
@app.callback(
    Output('Province/Territory_ENG', 'figure'),
    Output('Canada_ENG', 'figure'),
    Input('Engineering_Occupation', 'value')
)

def q3(occupation):

    # keeps graphs empty until value is selected
    if not occupation:
        return px.bar(title="Select at least one engineering occupation."), px.bar()

    # prep df to store the filtered data
    filtered_df = pd.DataFrame()

    # filter through the specific engineering occupation, create a new column and add said data to the new df (filtered_df)
    for keyword in occupation:
        df = master_df[master_df['Occupation'].str.contains(keyword, case=False, regex=True)].copy()
        df['Occupation Type'] = keyword.capitalize() + " Engineers"
        filtered_df = pd.concat([filtered_df, df])

    # Split into Canada and Provinces/Territories
    df_canada = filtered_df[filtered_df['Province'] == 'Canada']
    df_provinces = filtered_df[filtered_df['Province'] != 'Canada']

    # Combine engineering totals for Province/Territory by region and types
    prov_terr_eng = df_provinces.groupby(['Province', 'Occupation Type'])['Total'].sum().reset_index()
    # creates a horizontal bar graph for the Province/Territory
    fig_prov_terr = px.bar(prov_terr_eng, y='Province', x='Total', color='Occupation Type', orientation='h', title="Number of Engineers by Province")
    # format layout for organization
    fig_prov_terr.update_layout(title_x=0.5, title_font_size=20, showlegend=False)

    # Combine engineering totals for Canada by region and types
    can_eng = df_canada.groupby(['Occupation Type'])['Total'].sum().reset_index()
    # creates a bar graph for Canada
    fig_can = px.bar(can_eng, x='Occupation Type', y='Total', color='Occupation Type', title="Number of Engineers for All of Canada")
    # format layout for organization
    fig_can.update_layout(title_x=0.5, title_font_size=20)

    return fig_prov_terr, fig_can


# --------------------------------------------------------------------------------------------
# Callback and Graph - QUESTION 4
# --------------------------------------------------------------------------------------------
# callback for q4 graph
@app.callback(
    Output('Top_Occupation', 'figure'),
    Input('Province/Territory_Top', 'value'),
    Input('Slider', 'value')
)

def q4(region, top):

    # filter data by region
    df = master_df[master_df['Province'] == region]
    # group by NOC code (0-9)
    NOC = df.groupby('NOC')['Total'].sum().reset_index()
    # sort in decesending order based on total and choose the top x rows (for the slider) 
    NOC = NOC.sort_values(by='Total', ascending=False).head(top)

    # Map NOC codes to corresponding occupation
    Occupations = {
        '0': "0 Legislative and senior management occupations",
        '1': "1 Business, finance and administration occupations",
        '2': "2 Natural and applied sciences and related occupations",
        '3': "3 Health occupations",
        '4': "4 Education, law, and social services",
        '5': "5 Art, culture, and recreation",
        '6': "6 Sales and service",
        '7': "7 Trades and transport",
        '8': "8 Natural resources and agriculture",
        '9': "9 Manufacturing and utilities"
    }

    # add a new column with occupation labels
    NOC['Occupation'] = NOC['NOC'].map(Occupations)

    # creates a pie chart 
    fig = px.pie(NOC, names='Occupation', values='Total', title=f"Top {top} NOC Occupations in {region}")
    # format layout for organization
    fig.update_layout(title_x=0.5)
    
    return fig


# run the app
if __name__ == '__main__':
    app.run(debug=True)
```
